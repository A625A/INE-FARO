## Exploración inicial de ENCOVI 2023

Este notebook realiza un análisis exploratorio de los microdatos de ENCOVI 2023. Se cargan las bases de Personas y Hogares, se verifican las columnas principales y se revisa la estructura de los ponderadores (FACTOR / Factor3).

**Objetivo:** Generar un vistazo rápido a las dimensiones de los datos, valores faltantes y estadísticas descriptivas por departamento, sentando las bases para el cálculo de indicadores de informalidad y pobreza.

> **Importante:** Ejecuta este notebook desde la carpeta raíz del proyecto `INE-DATATON` (abre tu terminal o PowerShell, escribe `cd C:\Users\jptor\Documents\Proyectos JP\INE-DATATON` y luego lanza `jupyter notebook`). Si lo abres desde otra carpeta, las rutas relativas no encontrarán los archivos.

```python
import pandas as pd
import pyreadstat
import os
import warnings

warnings.filterwarnings('ignore')

# --- IMPRIMIR PARA DEPURACIÓN ----
print('--- Información de rutas ---')
print('Directorio de trabajo actual (cwd):', os.getcwd())
BASE_DIR = os.path.dirname(os.path.dirname(os.path.abspath(__file__)))
print('BASE_DIR (padre de la carpeta notebooks):', BASE_DIR)
RUTA_ENCOVI = os.path.join(BASE_DIR, 'data', 'raw', 'encovi_2023')
personas_path = os.path.join(RUTA_ENCOVI, 'encovi_2023_personas.sav')
print('Ruta a persons file:', personas_path)
print('¿El archivo existe?', os.path.exists(personas_path))
print('--- Fin de verificación ---')

# Cargar base de personas
try:
    df_personas, meta = pyreadstat.read_sav(personas_path)
    print(f'✅ Base de Personas cargada: {df_personas.shape[0]} filas x {df_personas.shape[1]} columnas')
except Exception as e:
    print(f'❌ Error al cargar base de personas: {e}')
    import traceback
    traceback.print_exc()
```

### Muestra de datos

A continuación, mostramos las primeras 5 filas de la base de personas para verificar que la carga fue correcta y que los tipos de datos son los esperados.

```python
display(df_personas.head())
```

### Estadísticas descriptivas

Estadísticas básicas de las variables numéricas clave, filtrando por el factor de expansión FACTOR.

```python
df_num = df_personas.select_dtypes(include=['float64', 'int64'])
print(df_num.describe().transpose())
```

### Verificación de ponderadores

Comparación rápida entre los dos factores de expansión (FACTOR y Factor3) para detectar posibles diferencias de ponderación según dominio o tipo de tabulación.

```python
print('Conteo de FACTOR no nulo:', df_personas['FACTOR'].notna().sum())
print('Conteo de Factor3 no nulo:', df_personas['Factor3'].notna().sum())
print('Correlación Pearson entre FACTOR y Factor3:', df_personas['FACTOR'].corr(df_personas['Factor3']))
```
